# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import 
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image using OpenCV
image=cv2.imread("baseball.jpg")
```

```python
# Display the image using Matplotlib
plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
plt.title("Original Image")
```

```python
tx, ty = 50, 80
M_translation = np.float32([[1, 0, tx],[0, 1, ty]])
translated_image = cv2.warpAffine(image, M_translation,(image.shape[1], image.shape[0]))
```

```python
plt.figure(figsize=(15,10))
plt.subplot(1,2,1);plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB));plt.title("Original image");
plt.subplot(1,2,2);plt.imshow(cv2.cvtColor(translated_image,cv2.COLOR_BGR2RGB));plt.title("Translated image");
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
rows,cols=image.shape[:2]
center=(cols//2,rows//2)
M45 = cv2.getRotationMatrix2D(center,45,1.0)
rotated_45 = cv2.warpAffine(image,M45,(cols,rows))

M90 = cv2.getRotationMatrix2D(center,90,1.0)
rotated_90 = cv2.warpAffine(image,M90,(cols,rows))
```

```python
plt.figure(figsize=(15,10))
plt.subplot(1, 3, 1); plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB)); plt.title("Original Image")
plt.subplot(1, 3, 2); plt.imshow(cv2.cvtColor(rotated_45,cv2.COLOR_BGR2RGB)); plt.title("45° Rotated Image")
plt.subplot(1, 3, 3); plt.imshow(cv2.cvtColor(rotated_90,cv2.COLOR_BGR2RGB)); plt.title("90° Rotated Image")
```
